# Phase 1: Mathematical Reasoning Dataset Exploration

Load the local HERCULEAN XBRL mathematical reasoning CSV and inspect its size, shape, exact source column names, first five rows, missing values, and unique values.

```python
from pathlib import Path
import pandas as pd

data_path = Path('../data/raw/mathematical_reasoning.csv')
df = pd.read_csv(
    data_path,
    dtype={
        'entity.cik': 'string',
        'report.accession': 'string',
        'report.id': 'string',
        'report.sic-code': 'string',
    },
    keep_default_na=False,
)
```

```python
print(f'File size: {data_path.stat().st_size:,} bytes')
print(f'Rows: {df.shape[0]:,}')
print(f'Columns: {df.shape[1]:,}')
print('Exact column names:')
for column_name in df.columns:
    print(f'  {column_name}')

print('First 5 rows:')
display(df.head(5))

print('Missing values by column:')
display(df.isna().sum().rename('missing_values').to_frame())

print('Unique values by column:')
display(df.nunique(dropna=False).rename('unique_values').to_frame())
```

## Phase I Findings

The local file contains XBRL Data Quality Committee (DQC) mathematical-consistency assertion occurrences. The dataset documentation describes `mathematical_reasoning.csv` as DQC ground-truth assertions generated over SEC EDGAR XBRL filings; this local file has 362 rows. A row is therefore an assertion occurrence, not a model prediction or a complete agent evaluation case. The row links assertion information to an entity and SEC report through fields such as `assertion.code`, `assertion.detail`, `entity.name`, `report.id`, `report.accession`, and `report.filing-year`.

```python
# Reproducible Phase I analysis using only columns present in the CSV.
print('Data types:')
display(df.dtypes.rename('dtype').to_frame())

print('Missing values:')
display(df.isna().sum().rename('missing_values').to_frame())

unique_columns = [
    'assertion.code',
    'assertion.type',
    'assertion.severity',
    'entity.name',
    'report.id',
    'report.filing-year',
]
print('Requested unique-value counts:')
display(df[unique_columns].nunique(dropna=False).rename('unique_values').to_frame())

categorical_columns = [
    'assertion.type',
    'assertion.severity',
    'report.filing-year',
    'report.document-type',
    'report.base-taxonomy',
]
for column_name in categorical_columns:
    print(f'Frequency table: {column_name}')
    display(df[column_name].value_counts(dropna=False).rename('count').to_frame())

print('Top 10 assertion codes:')
display(df['assertion.code'].value_counts(dropna=False).head(10).rename('count').to_frame())

print('Top 10 entities by assertion count:')
display(df['entity.name'].value_counts(dropna=False).head(10).rename('assertion_count').to_frame())

print('Assertions by type and filing year:')
display(pd.crosstab(df['assertion.type'], df['report.filing-year']))

print('Assertions by severity and filing year:')
display(pd.crosstab(df['assertion.severity'], df['report.filing-year']))

print('Assertions by entity:')
display(df.groupby('entity.name', dropna=False).agg(
    assertion_count=('assertion.code', 'size'),
    assertion_code_count=('assertion.code', 'nunique'),
    report_count=('report.id', 'nunique'),
).sort_values('assertion_count', ascending=False))

print('Potential dashboard fields:')
print({
    'filters': ['assertion.code', 'assertion.type', 'assertion.severity', 'entity.name', 'report.filing-year', 'report.document-type', 'report.base-taxonomy'],
    'charts': ['assertion.type', 'assertion.severity', 'report.filing-year', 'entity.name', 'assertion.code'],
    'case_detail_table': ['assertion.code', 'assertion.detail', 'entity.name', 'report.accession', 'report.id', 'report.filing-date', 'report.filing-year', 'report.document-type', 'report.sec-url'],
})

print('Agent-evaluation fields absent from this CSV: model/framework predictions, case-level pass/fail outcomes, cost, and latency.')
```

## HERCULEAN Table 2: Auditing Benchmark Results

The next cells transcribe only the Auditing block from Table 2 of the official HERCULEAN paper. Values are percentages and are kept in the paper's framework/model configuration order.

```python
benchmark_path = Path('../data/raw/benchmark_results.csv')
benchmark_df = pd.read_csv(benchmark_path)
metric_columns = ['acc', 'ser', 'eer', 'cer']
row_totals = benchmark_df[metric_columns].sum(axis=1)

assert list(benchmark_df.columns) == ['framework', 'model', 'acc', 'ser', 'eer', 'cer']
assert len(benchmark_df) == 20
assert benchmark_df[['framework', 'model']].drop_duplicates().shape[0] == 20
assert (row_totals.sub(100).abs() < 0.01).all()
assert ((benchmark_df[metric_columns] >= 0) & (benchmark_df[metric_columns] <= 100)).all().all()
assert benchmark_df['acc'].max() == 66.15

print('Validation passed')
print(f'Rows: {len(benchmark_df)}')
print(f'Maximum ACC: {benchmark_df["acc"].max():.2f}')
print('Maximum absolute row-sum difference from 100:', row_totals.sub(100).abs().max())

print('Summary by framework:')
display(benchmark_df.groupby('framework')[metric_columns].agg(['mean', 'max']).round(2))

print('Summary by model:')
display(benchmark_df.groupby('model')[metric_columns].agg(['mean', 'max']).round(2))
```

## Module 6 Cleaning and Verification

Before running the pipeline, review `cleaning_decision_log.md` for the proposed rules and the five predeclared record expectations. The pipeline checks the full DQC source, preserves the original file, exports a cleaned sample of at most 50 rows, validates the separate published Table 2 aggregate metrics, and writes the assignment evidence and PDF draft. It does not merge assertion occurrences with model outcomes.

```python
import json
import subprocess
import sys
from pathlib import Path

import pandas as pd

project_root = Path.cwd()
if not (project_root / 'src' / 'cleaning' / 'clean_project_data.py').exists():
    project_root = project_root.parent

run = subprocess.run(
    [sys.executable, str(project_root / 'src' / 'cleaning' / 'clean_project_data.py')],
    cwd=project_root,
    check=True,
    capture_output=True,
    text=True,
)
print(run.stdout.splitlines()[-1])

summary = json.loads((project_root / 'outputs/cleaning_summary.json').read_text(encoding='utf-8'))
print('DQC records / physical lines:', summary['dqc']['record_count'], '/', summary['dqc']['physical_line_count'])
print('Exact duplicate rows:', summary['dqc']['exact_duplicate_rows'])
print('Repeated report.id groups / excess occurrences:', summary['dqc']['repeated_report_id_groups'], '/', summary['dqc']['excess_repeated_report_id_records'])
print('Missing values before / after:', summary['dqc']['missing_before_total'], '/', summary['dqc']['missing_after_total'])
print('Five record checks passed:', summary['record_checks']['passed'], '/', summary['record_checks']['total'])

print('Top published aggregate configurations:')
ranked_results = pd.read_csv(project_root / 'outputs/ranked_configurations.csv')
display(ranked_results.head(10))

print('Individual record checks:')
record_checks = pd.read_csv(project_root / 'outputs/record_checks.csv')
display(record_checks)

print('Cleaned sample shape:')
cleaned_sample = pd.read_csv(project_root / 'outputs/cleaned_sample.csv', dtype='string', keep_default_na=False)
print(cleaned_sample.shape)
print('Draft PDF:', project_root / 'outputs/module6_cleaning_draft.pdf')
```
